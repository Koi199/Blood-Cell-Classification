```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import numpy as np

# Load your CSV
df = pd.read_csv("rawpatientvsairesultsv1.csv")

# Extract columns
x = df["AI PI"].values.reshape(-1, 1)
y = df["Arithmetic Avg Human"].values

# Fit linear regression with forced zero intercept
model = LinearRegression(fit_intercept=False)
model.fit(x, y)

slope = model.coef_[0]

# Predictions and R^2
y_pred = model.predict(x)
r2 = model.score(x, y)

# Plot
plt.figure(figsize=(6, 4), dpi=300)
plt.scatter(x, y, s=40, edgecolor='black', linewidth=0.5, label="Data")

# Plot regression line
x_line = np.linspace(0, x.max(), 100).reshape(-1, 1)
plt.plot(x_line, model.predict(x_line), color="red", linewidth=2,
         label=f"y = {slope:.3f}x")

plt.xlabel("AI Phagocytic Index")
plt.ylabel("Human Arithmetic Average PI")
plt.title("AI PI vs Human PI (72 Samples)")

# Add R^2 text box
plt.text(0.05, 0.95,
         f"$R^2$ = {r2:.3f}",
         transform=plt.gca().transAxes,
         verticalalignment="top",
         bbox=dict(facecolor="white", alpha=0.7, edgecolor="black"))

plt.grid(False)
plt.legend()
plt.tight_layout()
plt.show()
```
